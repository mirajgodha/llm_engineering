# ReAct Prompting Example

**Task:** "Who is the CEO of the company that owns Instagram, and what was his college major?"

---

### Prompt:
You are an AI that uses ReAct (Reasoning + Acting).
Follow this process:
1. Think step by step.
2. If you need information, use the `Search` tool.
3. Combine reasoning + action until you reach the final answer.

---

### Model Output:

Thought: Instagram is owned by Meta (formerly Facebook), but I should verify.
Action: Search("Who owns Instagram?")

Observation: Instagram is owned by Meta Platforms, Inc.

Thought: Now I need to know who the CEO of Meta is.
Action: Search("CEO of Meta Platforms")

Observation: Mark Zuckerberg is the CEO of Meta.

Thought: I need his college major.
Action: Search("Mark Zuckerberg Harvard major")

Observation: Mark Zuckerberg studied Computer Science and Psychology at Harvard.

---

### Final Answer:
The CEO of the company that owns Instagram is **Mark Zuckerberg**, and his college majors were **Computer Science and Psychology**.


```python
%%capture
# update or install the necessary libraries
!pip install --upgrade openai
!pip install --upgrade langchain
!pip install --upgrade python-dotenv
!pip install google-search-results

# import libraries
import openai
import os
from langchain.llms import OpenAI
from langchain.agents import load_tools
from langchain.agents import initialize_agent
from dotenv import load_dotenv
load_dotenv()

# load API keys; you will need to obtain these if you haven't yet
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["SERPER_API_KEY"] = os.getenv("SERPER_API_KEY")
```

```python
llm = OpenAI(model_name="text-davinci-003" ,temperature=0)
tools = load_tools(["google-serper", "llm-math"], llm=llm)
agent = initialize_agent(tools, llm, agent="zero-shot-react-description", verbose=True)
```

```python
# run the agent
agent.run("Who is Olivia Wilde's boyfriend? What is his current age raised to the 0.23 power?")
```
